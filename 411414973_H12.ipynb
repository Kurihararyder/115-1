```python
def a(total):
    print(f"原價金額：{total:g}元")
    print(f"9折金額：{total * 0.9:g}元")

def b(p, q):
    total = p * q
    a(total)

def main():
    p = float(input("請輸入單價："))
    q = int(input("請輸入數量："))
    b(p, q)

main()
```
