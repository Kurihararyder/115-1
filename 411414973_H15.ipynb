```python
listA = []
while True:
    item = input("請輸入A館文具，輸入'/'結束：")
    if item == "/":
        break
    listA.append(item)

listB = []
while True:
    item = input("請輸入B館文具，輸入'/'結束：")
    if item == "/":
        break
    listB.append(item)

listC = listA + listB
listC.append("膠水")
listC = "/".join(listC)

print(f"二館合併清單：{listC}")
```
