```python
order = []
print("請輸入您要點的餐點，輸入'/'結束：")
while True:
    item = input()
    if item == "/":
        break
    order.append(item)

if "沙拉" in order:
    order.remove("沙拉")
    print("今日無沙拉，已自動移除。")

order.append("冰淇淋")

print("完整訂單清單：")
print(", ".join(order))
```
