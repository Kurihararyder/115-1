```python
height=float(input("請輸入身高："))
weight=float(input("請輸入體重："))
height=height/100

bmi = weight/(height ** 2)

if bmi<18.5:
    status="過輕"
elif bmi<24:
    status="正常"
else:
    status="過重"

print(f"你的BMI值為:{bmi:.2f}，判定為：{status}")
```
