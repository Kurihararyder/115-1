```python
import random



day = 1
win = False
while day <= 3:
    g = random.randint(1,50)
    h = int(input(f"今天第{day}天,請輸入1-50數字: "))
    print(f"今天的幸運數字是{g}")
    if g == h:
        print("恭喜你，中獎了！")
        win = True
        break
    else:
        print("你沒中獎，明天再來\n")
    day += 1
if not win:
    print("很遺憾3天都沒中獎，下次再試試吧！")
```
